## 2. Dataset source

The Gapminder data were compiled from international statistics and are available as a public CSV file through Plotly's datasets repository. Each case (row) represents one country in one year, with information on population, continent, life expectancy, and GDP per capita. The data cover 142 countries in 12 years, from 1952 to 2007.

The data file is available at [Plotly's Gapminder dataset on GitHub](https://github.com/plotly/datasets/blob/master/gapminderDataFiveYear.csv). It can be downloaded for free without registration.
